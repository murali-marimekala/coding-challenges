```python
# Import the TwoSum solution (if needed)
# You can copy your solution here or import it

def twoSum(nums, target):
    """
    Find two numbers that add up to target.
    Returns indices of the two numbers.
    """
    complement_map = {}
    
    for i, num in enumerate(nums):
        complement = target - num
        if complement in complement_map:
            return [complement_map[complement], i]
        complement_map[num] = i
    
    return []
```

```python
# Test Case 1
nums1 = [2, 7, 11, 15]
target1 = 9
result1 = twoSum(nums1, target1)
print(f"Test 1: nums = {nums1}, target = {target1}")
print(f"Result: {result1}")
print(f"Values: {nums1[result1[0]]}, {nums1[result1[1]]}")
print()
```

## Step-by-Step Visualization

Watch the algorithm execute line by line and see how the hash map builds up:


```python
def twoSum_verbose(nums, target):
    """
    Same algorithm but prints each step
    """
    print(f"🎯 Finding two numbers that sum to {target}")
    print(f"📊 Input array: {nums}\n")
    
    complement_map = {}
    print("=" * 60)
    print("STEP-BY-STEP EXECUTION")
    print("=" * 60)
    
    for i, num in enumerate(nums):
        complement = target - num
        
        print(f"\n📍 Step {i+1}: Looking at number {num} (index {i})")
        print(f"   ❓ Need complement: {target} - {num} = {complement}")
        print(f"   📦 Current map: {complement_map}")
        
        if complement in complement_map:
            result_idx = complement_map[complement]
            print(f"   ✅ FOUND! {complement} exists in map at index {result_idx}")
            print(f"   🎉 Answer: indices [{result_idx}, {i}] with values [{nums[result_idx]}, {num}]")
            print(f"   ✓ Verification: {nums[result_idx]} + {num} = {nums[result_idx] + num}")
            return [result_idx, i]
        
        complement_map[num] = i
        print(f"   ➕ Added {num} → index {i} to map")
    
    print(f"\n❌ No solution found!")
    return []

# Example: Run with first test case
print("=" * 60)
twoSum_verbose([2, 7, 11, 15], 9)
```

```python
# Try the verbose version with different test cases
print("\n\n")
print("🧪 TEST CASE 2:")
twoSum_verbose([3, 2, 4], 6)

print("\n\n")
print("🧪 TEST CASE 3:")
twoSum_verbose([3, 3], 6)
```

## Algorithm Execution Table

See how the map grows at each iteration:


```python
import pandas as pd

def twoSum_table_visualization(nums, target):
    """
    Execute algorithm and show state as a table
    """
    print(f"\n🎯 Target: {target}")
    print(f"📊 Array: {nums}\n")
    
    steps_data = []
    complement_map = {}
    
    for i, num in enumerate(nums):
        complement = target - num
        found = complement in complement_map
        
        steps_data.append({
            'Step': i + 1,
            'Current#': num,
            'Index': i,
            'Need': complement,
            'InMap?': '✅ YES' if found else '❌ NO',
            'Map State': str(complement_map.copy())
        })
        
        if found:
            result_idx = complement_map[complement]
            print(f"✨ SOLUTION FOUND at step {i+1}!")
            print(f"   Indices: [{result_idx}, {i}]")
            print(f"   Values: {nums[result_idx]} + {num} = {target}")
            break
        
        complement_map[num] = i
    
    df = pd.DataFrame(steps_data)
    display(df)

# Visualize with Example 1
print("=" * 70)
print("EXAMPLE 1: nums = [2, 7, 11, 15], target = 9")
print("=" * 70)
twoSum_table_visualization([2, 7, 11, 15], 9)

print("\n\n")
print("=" * 70)
print("EXAMPLE 2: nums = [3, 2, 4], target = 6")
print("=" * 70)
twoSum_table_visualization([3, 2, 4], 6)
```
